```python
# ============================================================
# WEATHER & SOIL-BASED CROP RECOMMENDATION SYSTEM
# DATA ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# ------------------------------------------------------------
# 1. Load Dataset
# ------------------------------------------------------------

BASE_DIR = Path.cwd().parent
DATA_PATH = BASE_DIR / "data" / "crop_data.csv"

crop_df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Dataset path:", DATA_PATH)

# ------------------------------------------------------------
# 2. Display Dataset
# ------------------------------------------------------------

print("\nFirst 5 rows:")
print(crop_df.head())

# ------------------------------------------------------------
# 3. Dataset Information
# ------------------------------------------------------------

print("\nDataset Shape:")
print("Rows:", crop_df.shape[0])
print("Columns:", crop_df.shape[1])

print("\nColumn Names:")
print(list(crop_df.columns))

print("\nDataset Information:")
crop_df.info()

# ------------------------------------------------------------
# 4. Statistical Summary
# ------------------------------------------------------------

print("\nStatistical Summary:")
print(crop_df.describe())

# ------------------------------------------------------------
# 5. Missing Values
# ------------------------------------------------------------

print("\nMissing Values:")
print(crop_df.isnull().sum())

# ------------------------------------------------------------
# 6. Duplicate Values
# ------------------------------------------------------------

print("\nDuplicate Rows:")
print(crop_df.duplicated().sum())

# ------------------------------------------------------------
# 7. Crop Classes
# ------------------------------------------------------------

print("\nNumber of Crop Classes:")
print(crop_df["label"].nunique())

print("\nCrop Classes:")
print(sorted(crop_df["label"].unique()))

# ------------------------------------------------------------
# 8. Crop Distribution
# ------------------------------------------------------------

crop_counts = crop_df["label"].value_counts()

print("\nCrop Distribution:")
print(crop_counts)

plt.figure(figsize=(10, 5))
crop_counts.plot(kind="bar")
plt.title("Crop Distribution")
plt.xlabel("Crop")
plt.ylabel("Number of Samples")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 9. Numerical Features
# ------------------------------------------------------------

features = [
    "N",
    "P",
    "K",
    "temperature",
    "humidity",
    "ph",
    "rainfall"
]

print("\nNumerical Features:")
print(features)

# ------------------------------------------------------------
# 10. Feature Distributions
# ------------------------------------------------------------

crop_df[features].hist(
    figsize=(14, 10),
    bins=10
)

plt.suptitle("Distribution of Soil and Weather Features")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 11. Correlation Analysis
# ------------------------------------------------------------

correlation = crop_df[features].corr()

print("\nCorrelation Matrix:")
print(correlation)

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Between Soil and Weather Parameters")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 12. Average Parameters by Crop
# ------------------------------------------------------------

crop_summary = crop_df.groupby("label")[features].mean()

print("\nAverage Soil and Weather Parameters by Crop:")
print(crop_summary)

# ------------------------------------------------------------
# 13. Dataset Quality Check
# ------------------------------------------------------------

print("\n" + "=" * 50)
print("DATASET QUALITY CHECK")
print("=" * 50)

print("Total samples:", len(crop_df))
print("Total features:", len(features))
print("Total crop classes:", crop_df["label"].nunique())
print("Total missing values:", crop_df.isnull().sum().sum())
print("Total duplicate rows:", crop_df.duplicated().sum())

# ------------------------------------------------------------
# 14. Final Summary
# ------------------------------------------------------------

print("\n" + "=" * 50)
print("DATA ANALYSIS COMPLETED")
print("=" * 50)

print("The crop dataset has been successfully analyzed.")
print("The dataset is ready for the preprocessing stage.")
```
